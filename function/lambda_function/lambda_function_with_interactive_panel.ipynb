```python
#lambda function  pregram number_add

number_add = lambda n1,n2 : n1+n2
print(number_add(100,200))
```

```python
#lambda function for substracting number

substracting_number = lambda n1,n2 : n1 - n2
print(substracting_number(100,300))
```

```python
#lamda function for max_number

max_two = lambda n1,n2:n1 if n1> n2 else n2
print(max_two(100,200))
```

```python
#min of two

min_two = lambda n1,n2: n1 if n1<n2 else n2
print(min_two(100,200))
```
