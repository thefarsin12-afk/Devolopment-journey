Shapes

```python
#Rectangle:Multiply length (l) by width (w) (A = l × w)

class Shapes:

    name : str

    def __init__(self,name):

        self.name = name

class Rectangile(Shapes):

    length : int

    width : int

    def __init__(self,name,length,width):

        super().__init__(name)

        self.length = length

        self.width = width

    def result(self):

        print(self.name, "=" ,self.length * self.width)  


rectangile_instant1 = Rectangile("Rectangile",20,12)

rectangile_instant1.result()
```

```python

#Square: Multiply the side length (a) by itself (A = a²)

class Square(Shapes):

    length : int

    def __init__(self,name,length):

        super().__init__(name)

        self.length = length

    def result(self):

        print(self.name,self.length **2)

square_instant1 = Square("Square = ",10)
square_instant1.result()
```

```python
#Circle: Multiply pi (π) by the radius squared (A = π r²)

class Circle(Shapes):

    radius : float

    def __init__(self,name,radius):

        super().__init__(name)

        self.radius = radius

    def result(self):

        print("area",self.name,"=",3.14*(self.radius**2))   

circle_instant1 = Circle("Circle",12)
circle_instant1.result()
```

```python
#Parallelogram: Multiply base (b) by vertical height (h) (A = bh).

class Parallelogram(Shapes):

    base = int

    height = int

    def __init__(self,name,base,height):

        super().__init__(name)

        self.base = base

        self.height = height

    def result(self):

        print("area", self.name,"=",self.base * self.height) 

parallelogram_instatant = Parallelogram("Parallelogram",12,20)
parallelogram_instatant.result()           


    
```

```python
#Trapezium: Add parallel sides (a + b), multiply by height (h), 
# and divide by two (\(A = \frac{1}{2}(a + b)h\)).

class Trapezium(Shapes):

    a : float

    b : float

    h : float

    def __init__(self,name,a,b,h):

        super().__init__(name)

        self.a = a

        self.b = b

        self.h = h

    def result(self):

        print("Result",self.name,"=",((self.a + self.b )*self.h) / 2)  

trapezium_instant1 = Trapezium("Trapezium",12,20,3)
trapezium_instant1.result()          
```
