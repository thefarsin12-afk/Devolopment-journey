##Movie Case Study


```python
movies = [

    ["kgf","kannada",150,2005,8],
    ["balan","malayalam",130,2026,7],
    ["ramayan","hindi",150,2026,8.5],
    ["abcd","malayalam",140,2008,6],
    ["goatlife","malayalam",160,2024,9]
]
```

```python
all_mobie_title = {m [0] for m in movies}
all_mobie_title
```

```python
movies_languages = {m [1] for m in movies}
movies_languages
```

```python
movie_uppercase = {m[0].upper() for m in movies}
movie_uppercase
```

```python
name_filter = [m[0] for m in movies if m[0].startswith("a")]
name_filter
```

```python
#display malayalam movies,filter

malayala_movies = [m[0] for m in movies if m[1] =="malayalam"]
malayala_movies
```

```python
#language wise count movies

languages = [m[1] for m in movies]

lc = {l: languages.count(l) for l in languages}
lc
```

```python
import matplotlib.pyplot as plt
x = lc.keys()
y = lc.values()
plt.bar(x,y)
```

```python
ratings = {

"maximum" : max(m[4] for m in movies),

"minimum" :min(m[4] for m in movies),

"average" :sum(m[4] for m in movies) / len(movies)
}
```

```python
import matplotlib.pyplot as plt

x = ratings.keys()
y = ratings.values()
plt.bar(x,y)

```
